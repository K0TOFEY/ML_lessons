```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

file_path = "ПЗ 2. LinReg.xlsx"
```

```python
df1 = pd.read_excel(file_path, 'LinReg1')
df1
```

```python
X1 = df1.iloc[:, [0]]
y1 = df1.iloc[:, [1]]

model1 = LinearRegression()
model1.fit(X1, y1)

a1 = model1.coef_[0]
b1 = model1.intercept_

y1_pred = model1.predict(X1)
result1 = df1.copy()
result1["Y_pred"] = y1_pred
result1["Error"] = result1.iloc[:, 1] - result1["Y_pred"]

result1
```

```python
def regression_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    return {"MAE": f"{mae:.2f}", 
            "MSE": f"{mse:.2f}", 
            "RMSE": f"{rmse:.2f}", 
            "R2": f"{r2:.2f}"
            }

print(regression_metrics(y1, y1_pred))
```

```python
plt.figure(figsize=(8, 5)) # Полотно figsize(n, m) - где n, m умножается на 100пикселей
plt.scatter(X1, y1, label="Исходные точки")
plt.plot(X1, y1_pred, label="Линия регрессии")
plt.xlabel("Масса")
plt.ylabel("Сила тяжести")
plt.title("LinReg1 — линейная регрессия")
plt.legend()
plt.grid()
plt.show()
```
