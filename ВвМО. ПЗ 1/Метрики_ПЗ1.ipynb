```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
file_path = 'ПЗ1.Метрики.xlsx'

df_reg = pd.read_excel(file_path, sheet_name='Regression')
df_bin = pd.read_excel(file_path, sheet_name='Binary Classification')
df_reg
```

```python
def reg_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    return {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    }

reg_rows = []
for (model, part), group in df_reg.groupby(["model", "part"]):
    result = reg_metrics(group["Y_true"].values, group["Y_pred"].values)

    result["model"] = model
    result["part"] = part

    reg_rows.append(result)

print(result)
reg_result = pd.DataFrame(reg_rows)[['model', 'part', 'MAE', 'MSE', 'RMSE', 'R2']]
reg_result = reg_result.sort_values(["model", "part"]).reset_index(drop=True)
reg_result
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score
```

```python
def bin_metrics(y_true, y_pred, y_proba):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "F1-score": f1_score(y_true, y_pred, zero_division=0),
        "AUC ROC": roc_auc_score(y_true, y_proba),
        "AUC PR": average_precision_score(y_true, y_proba)
    }

bin_rows = []
bin_curves = {}
cm_dict = {}

for (model, part), group in df_bin.groupby(["model", "part"]):
    y_true = group["Y_true"].values
    y_pred = group["Y_pred"].values
    y_proba = group["Y_pred_proba_1"].values

    m = bin_metrics(y_true, y_pred, y_proba)
    m["model"] = model
    m["part"] = part
    bin_rows.append(m)
    # print(bin_rows)

bin_results = pd.DataFrame(bin_rows)[
    ["model", "part", "Accuracy", "Precision", "Recall", "F1-score", "AUC ROC", "AUC PR"]
].sort_values(["model", "part"]).reset_index(drop=True)

bin_results
```
